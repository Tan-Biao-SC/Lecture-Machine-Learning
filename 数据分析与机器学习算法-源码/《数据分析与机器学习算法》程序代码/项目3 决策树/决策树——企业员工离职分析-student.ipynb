## ☞ 1. 决策树的 what、where、when

☞☞ what——从离职讲起

><font color=orange size=3.5> 为达到目标 来 根据一定的条件进行选择，这个过程 就是决策树。

><font  size=3>员工👨‍💻：人事您好，我想要提出离职。</font>

><font  size=3>企业HR👩‍💼：为什么想要离职吗？能说说原因吗？是因为薪资问题吗？</font>

><font  size=3>员工👨‍💻：不是的，薪资很满意的。</font>

><font  size=3>企业HR👩‍💼：是因为工作时长太长了吗？</font>

><font  size=3>员工👨‍💻：不是的，工作时长在我的接受范围内。</font>

><font  size=3>企业HR👩‍💼： 那是因为工作中压力太大了吗？</font>

><font  size=3>员工👨‍💻：不是的，压力对于我来说属于中等水平。</font>

><font  size=3>企业HR👩‍💼： 那是因为...？</font>

><font  size=3>员工👨‍💻：...</font>

><font  size=3>经历过N轮问答后：</font>

><font  size=3>企业HR👩‍💼：请问是职位晋升问题吗？</font>

><font  size=3>员工👨‍💻：是的。</font>



><font  size=3.5>一般情况下，我们从顶部节点输入数据，经过多次判断后，这些数据就会被分为不同的类别。这就构成了一颗简单的分类决策树。


☞☞ 决策树算法的优缺点

><font  size=3.5>优点：1.直观、可视化、好理解 &emsp;&emsp;2、易于追溯和倒推
    
><font  size=3.5>缺点：1.容易发生结果不准确（过拟合） &emsp;&emsp;2、容易忽略数据中属性的相互关联

☞☞ where——哪里会用到决策树算法

><font color=orange size=3.5> 医学诊断 和 商业决策等 可视化需求较高的业务

☞☞ when——什么时候要用到决策树算法

><font color=orange size=3.5> 当需要处理有缺失属性的样本时、数据的特征不相关时、数据集较大且需要实时处理时。

## ✏️ 步骤3 构建决策树分类模型

&emsp;&emsp;<font size=3.5>数据准备完成后，现在需要建立模型了；这里是调用```sklearn```中的```DecisionTreeClassifier```函数创建决策树分类模型；</font>

<details>
  <summary><font size=3.5 >DecisionTreeClassifier函数的参数说明：</font></summary>
    
```python
class sklearn.tree.DecisionTreeClassifier(criterion=’gini’, splitter=’best’, max_depth=None,  min_samples_leaf=1, max_features=None, class_weight=None)
```
&emsp;&emsp;<font size=3.5>其中</font>

><font size=3.5>```criterion```：</font><font  size=3.5>可选参数，特征选择标准，有'gini'或者'entropy'两种特征选择方式； </font>

><font size=3.5 >```splitter```:  </font><font  size=3.5> 用来确定每个节点的分裂策略。支持 “最佳” 或者“随机”，数据集样本不大选择best，样本大最好选择random。</font>

><font size=3.5>```max_depth```:</font><font  size=3.5>  用来控制我们的人工智能针对某一特征连续追问的深度，如果过于刨根问底的追问有可能会让这棵决策树失智。</font>

><font size=3.5>```min_samples_leaf```:</font><font  size=3.5> 通过限制最大叶子节点数，数值越大，这棵决策树的情商会比较高，追问方式比较平滑。</font>

><font size=3.5 >```class_weight```:</font><font  size=3.5>  指定样本各类别的的权重，主要是为了防止训练集某些类别的样本过多导致训练的决策树过于偏向这些类别。设置为“balanced”，则决策树会自己计算权重，样本量少的类别所对应的样本权重会高</font>

><font size=3.5 >```random_state```:</font><font  size=3.5> 可选参数，默认是None。随机数种子。</font>
