## ☞ 1. 朴素贝叶斯

&emsp;&emsp;<font  size=3>朴素贝叶斯算法在是机器学习中核心算法之一，也经常用来解决现实生活中的问题；因为生活中绝大多数决策面临的信息都是不完全的，我们手中只有有限的信息。既然无法得到全面的信息，我们就应该在信息有限的情况下，尽可能做出一个最优的预测：我们生活中也经常用贝叶斯思维来做出决策。比如，你和客户在聊天的时候，如果对方说出“虽然”两个字，你大概就会猜测，对方后继九成的可能性会说出“但是”。我们的大脑可以根据生活的经历有了主观判断（先验概率），然后根据搜集新的信息来修正（调整因子），最后做出高概率的预测（后验概率）。<font  size=3>朴素贝叶斯模型被广泛应用于互联网新闻的分类、垃圾邮件的筛选等分类任务。</font>

☞☞ **思想由来**

&emsp;&emsp;<font  size=3>我们以一个例子来模拟朴素贝叶斯进行分类或决策的过程：</font>

><font  size=3>当我们在浏览网页时，看到“人工智能是否造成人类失业”这个信息；</font>

><font  size=3>你根据自己的理解（主观判断），做出了人工智能造成人类失业的概率为50%；</font>

><font  size=3>但是当你学习了一些数据分析，或者看了些这方面的最新研究进展（新的信息），然后你根据掌握的最新信息，自己对人工智能这个领域有了更深的理解；更加认为人工智能不可能导致人类事业，这时候你有了一个对之前得出的概率的一个调整因子，比如说0.8；</font>
    
><font  size=3>最后重新理解了“人工智能”这个信息（后验概率），认为人工智能能导致人类失业的概率只有40%。这也就是胡适说的“大胆假设，小心求证”。</font>

## ✏️ 步骤3 数据预处理

&emsp;&emsp;<font  size=3.5>数据采集完成以后，就要开始提取特征了，我们这里使用的是TF-IDF特征。TF-IDF是一种用以评估一个字词对于一个语料或者一个文本的重要程度，TF-IDF的主要思想是：如果某个词或短语在一篇文章中出现的频率TF高，并且在其他文章中很少出现，则认为此词或者短语具有很好的类别区分能力，适合用来分类；</font>

&emsp;&emsp;<font  size=3.5>使用Sklearn中的```Countvectorizer()```是将一个文档转换为向量，计算词汇在文本中出现的频率。它通过fit_transform函数计算各个词语出现的次数。</font>

&emsp;&emsp;<font  size=3.5>使用```TfidfTransformer```方法是将```Countvectorizer()```得到的结果进行词语重要程度计算，先计算Countvectorizer()，再计算TfidfTransformer</font>
