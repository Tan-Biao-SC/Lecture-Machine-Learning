# 项目 2｜用高斯朴素贝叶斯给鸢尾花分类

运行下面的代码，用训练集训练一个模型，再看看它把测试集的第一朵鸢尾花分到了哪个品种。

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB

iris = load_iris()
X_train, X_test, y_train, y_test = train_test_split(
    iris.data, iris.target, test_size=0.25, random_state=15
)

model = GaussianNB()
model.fit(X_train, y_train)
prediction = model.predict(X_test[:1])[0]

print("预测品种：", iris.target_names[prediction])
print("真实品种：", iris.target_names[y_test[0]])
```

`fit` 从训练集的测量值和品种标签中学习。
`predict` 根据测试样本的测量值给出预测品种。

## 拓展 1｜换一朵花

修改样本编号 `i`（0 到 37），再运行代码，比较预测品种和真实品种。

```python
i = 1
prediction = model.predict(X_test[i:i + 1])[0]

print("预测品种：", iris.target_names[prediction])
print("真实品种：", iris.target_names[y_test[i]])
```

## 拓展 2｜预测整批测试样本

用刚才训练好的模型预测整批测试样本。读一读代码，找出哪一行逐条比较预测品种和真实品种，并说出准确率的分母是什么。

```python
y_pred = model.predict(X_test)
correct_count = (y_pred == y_test).sum()
accuracy = correct_count / len(y_test)

print(f"预测正确：{correct_count}/{len(y_test)}")
print(f"准确率：{accuracy:.1%}")
```

## 拓展 3｜只量两项尺寸

数据的第 0、1 列是萼片长和萼片宽，第 2、3 列是花瓣长和花瓣宽。

先运行下一格，只用两项花瓣尺寸训练和预测。记下结果，再把 `[2, 3]` 改成 `[0, 1]`，用两项萼片尺寸重复实验。两次实验用的是同一批训练样本和测试样本。

```python
columns = [2, 3]
two_feature_model = GaussianNB()
two_feature_model.fit(X_train[:, columns], y_train)
two_feature_predictions = two_feature_model.predict(X_test[:, columns])

two_feature_correct = (two_feature_predictions == y_test).sum()
two_feature_accuracy = two_feature_correct / len(y_test)
print(f"特征列：{columns}")
print(f"预测正确：{two_feature_correct}/{len(y_test)}")
print(f"准确率：{two_feature_accuracy:.1%}")
```
